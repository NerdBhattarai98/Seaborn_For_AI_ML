**sns.histplot or sns.kdeplot (Distributions)**

What it does: Shows how a single numerical feature is distributed (e.g., checking if test scores or house prices are skewed or bell-shaped).

**sns.scatterplot or sns.lineplot (Relationships)**

What it does: Explores how two continuous numerical variables relate to each other (e.g., square footage vs. house price).

**sns.boxplot or sns.violinplot (Categorical + Statistical Spread)**

What it does: Compares a numerical value across different categories while instantly showing you the median, quartiles, and outliers (crucial for cleaning ML data).

**sns.countplot or sns.barplot (Categorical Comparisons)**

What it does: Counts how many rows belong to each category (like your Pokémon types) or calculates the average of a metric per category.

sns.heatmap (Correlations)

What it does: Displays a correlation matrix for your dataset. This is mandatory for Machine Learning because it helps you see which features are heavily correlated with each other before training a model.

sns.pairplot (The Ultimate Quick-Scan)

What it does: Automatically creates a grid of scatterplots and histograms for every single numerical column in your dataframe all at once. It's the ultimate shortcut for initial EDA.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load example dataset
df = sns.load_dataset("tips")

# Set the default graph style
sns.set_theme(style="whitegrid")


# ==========================================================
# 1. HISTOGRAM - sns.histplot()
# Definition: Shows the frequency distribution of numerical data.
# Why use it: To understand the spread, shape, and common ranges
# of values, and to identify skewness or unusual observations.
# Example use: Understanding the distribution of salaries or ages.
# ==========================================================

sns.histplot(data=df, x="total_bill", bins=20, kde=True)
plt.title("Distribution of Total Bills")
plt.show()


```

```python
# ==========================================================
# 2. SCATTER PLOT - sns.scatterplot()
# Definition: Displays the relationship between two numerical
# variables using individual points.
# Why use it: To identify relationships, clusters, and outliers.
# Example use: Checking whether higher bills are associated
# with higher tips.
# ==========================================================

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)
plt.title("Total Bill vs Tip")
plt.show()


```

```python
# ==========================================================
# 3. LINE PLOT - sns.lineplot()
# Definition: Connects values along an ordered axis to show
# how a quantity changes.
# Why use it: To identify trends and compare changes between
# groups over time or another ordered variable.
# Example use: Tracking sales over months or model accuracy
# across training epochs.
# Note: Repeated observations are aggregated by default.
# ==========================================================

sns.lineplot(
    data=df,
    x="size",
    y="total_bill",
    hue="sex",
    marker="o"
)
plt.title("Average Total Bill by Party Size")
plt.show()


```

```python

# ==========================================================
# 4. BOX PLOT - sns.boxplot()
# Definition: Summarizes numerical data using quartiles,
# the median, whiskers, and potential outliers.
# Why use it: To compare distributions and identify
# unusually high or low observations.
# Example use: Comparing salary distributions across departments.
# Note: Points beyond whiskers are potential outliers,
# not necessarily errors.
# ==========================================================

sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)
plt.title("Bill Distribution by Day")
plt.show()


```

```python
# ==========================================================
# 5. VIOLIN PLOT - sns.violinplot()
# Definition: Shows the estimated density and shape of a
# numerical distribution for one or more groups.
# Why use it: To compare distribution shapes, spread,
# and regions where values are concentrated.
# Example use: Comparing exam score distributions across classes.
# ==========================================================

sns.violinplot(
    data=df,
    x="day",
    y="total_bill",
    inner="quartile"
)
plt.title("Bill Distribution by Day")
plt.show()


```

```python
# ==========================================================
# 6. COUNT PLOT - sns.countplot()
# Definition: Displays the number of observations in each
# category using bars.
# Why use it: To understand category frequencies and find
# rare categories or imbalanced classes.
# Example use: Checking whether a classification dataset
# has many more samples of one class than another.
# ==========================================================

sns.countplot(
    data=df,
    x="day",
    hue="sex"
)
plt.title("Number of Records by Day and Sex")
plt.show()


```

```python
# ==========================================================
# 7. BAR PLOT - sns.barplot()
# Definition: Compares a numerical statistic, usually the mean,
# across different categories.
# Why use it: To summarize and compare numerical values
# between groups.
# Example use: Comparing average sales across regions.
# Note: Unlike countplot(), it summarizes a numerical variable
# instead of simply counting observations.
# ==========================================================

sns.barplot(
    data=df,
    x="day",
    y="tip"
)
plt.title("Average Tip by Day")
plt.show()


```

```python
# ==========================================================
# 8. REGRESSION PLOT - sns.regplot()
# Definition: Displays a scatter plot with a fitted regression
# line that summarizes a possible relationship.
# Why use it: To inspect a linear trend and see how observations
# deviate from the fitted line.
# Example use: Exploring the relationship between house size
# and house price before building a regression model.
# Note: Correlation does not prove causation.
# ==========================================================

sns.regplot(
    data=df,
    x="total_bill",
    y="tip"
)
plt.title("Regression: Total Bill vs Tip")
plt.show()


```

```python
# ==========================================================
# 9. HEATMAP - sns.heatmap()
# Definition: Represents values in a matrix using different colors.
# Why use it: To visualize numerical patterns and correlations
# between multiple features.
# Example use: Identifying strongly correlated features
# before training a machine learning model.
# Note: Correlation alone does not prove causation.
# ==========================================================

corr = df.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    center=0
)
plt.title("Feature Correlation Matrix")
plt.show()


```

```python
# ==========================================================
# 10. PAIR PLOT - sns.pairplot()
# Definition: Creates a grid of plots showing pairwise
# relationships between numerical variables.
# Why use it: To explore many feature relationships at once
# and discover patterns or clusters.
# Example use: Exploring features before selecting inputs
# for a machine learning model.
# Note: Best for a small number of numerical features because
# the plot can become crowded with many columns.
# ==========================================================

sns.pairplot(
    df,
    vars=["total_bill", "tip", "size"],
    hue="sex",
    diag_kind="hist"
)
plt.show()


```

```python
# ==========================================================
# 11. KDE PLOT - sns.kdeplot()
# Definition: Estimates a smooth probability density curve
# for numerical data using Kernel Density Estimation.
# Why use it: To understand distribution shape without
# dividing data into histogram bins.
# Example use: Comparing the distributions of two groups.
# Note: The curve depends on the smoothing bandwidth.
# ==========================================================

sns.kdeplot(
    data=df,
    x="total_bill",
    fill=True
)
plt.title("Density of Total Bills")
plt.show()


```

```python
# ==========================================================
# 12. JOINT PLOT - sns.jointplot()
# Definition: Combines a plot of two variables with separate
# distributions for each variable along the margins.
# Why use it: To understand the relationship between two
# variables and their individual distributions together.
# Example use: Exploring the relationship between advertising
# spending and sales.
# ==========================================================

sns.jointplot(
    data=df,
    x="total_bill",
    y="tip",
    kind="reg"
)
plt.show()
```
